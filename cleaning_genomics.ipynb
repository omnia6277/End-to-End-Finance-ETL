```python
#clean_balance_sheet
import pandas as pd
from sqlalchemy import create_engine, text
import re


server = "localhost"
database = "staging_genomics"

connection_string = (
    f"mssql+pyodbc://@{server}/{database}"
    "?driver=ODBC+Driver+18+for+SQL+Server"
    "&TrustServerCertificate=yes"
)
staging_engine = create_engine(connection_string)

server = "localhost"
database = "cleaning_genomics"

connection_string = (
    f"mssql+pyodbc://@{server}/{database}"
    "?driver=ODBC+Driver+18+for+SQL+Server"
    "&TrustServerCertificate=yes"
)
cleaning_engine = create_engine(connection_string)


# Clean data

def clean_financial_table(df_raw: pd.DataFrame, statement_type: str) -> pd.DataFrame:
    df = df_raw.copy()

    new_header = df.iloc[0].copy()
    
    new_header[0] = "Metric_Name"
    
    df.columns = new_header
    df = df.drop(index=[0, 1]).reset_index(drop=True)

    df = df.dropna(subset=["Metric_Name"])
    df = df[df["Metric_Name"].astype(str).str.strip() != ""].reset_index(drop=True)

    
    date_columns = [c for c in df.columns if c != "Metric_Name"]
    df_long = df.melt(
        id_vars=["Metric_Name"],
        value_vars=date_columns,
        var_name="Date",
        value_name="Value"
    )

    df_long["Date"] = pd.to_datetime(df_long["Date"], errors="coerce").dt.date
    df_long["Value"] = pd.to_numeric(df_long["Value"], errors="coerce").fillna(0.0)
    df_long.insert(0, "Statement_Type", statement_type)

    return df_long[["Statement_Type", "Metric_Name", "Date", "Value"]]


df_bs_raw = pd.read_sql("SELECT * FROM dbo.raw_balance_sheet", staging_engine)
print(f"Raw rows read: {len(df_bs_raw)}")

df_bs_clean = clean_financial_table(df_bs_raw, "Balance Sheet")

df_bs_clean.to_sql(
    name="clean_balance_sheet", con=cleaning_engine,
    if_exists="replace", index=False, schema="dbo"
)
print(f"Clean rows written: {len(df_bs_clean)}")
```

```python
#Combining
import pandas as pd
from sqlalchemy import create_engine, text
import re


server = "localhost"
database = "cleaning_genomics"

connection_string = (
    f"mssql+pyodbc://@{server}/{database}"
    "?driver=ODBC+Driver+18+for+SQL+Server"
    "&TrustServerCertificate=yes"
)
cleaning_engine = create_engine(connection_string)

df_bs = pd.read_sql("SELECT * FROM dbo.clean_balance_sheet",   cleaning_engine)
df_cf = pd.read_sql("SELECT * FROM dbo.clean_cash_flow",       cleaning_engine)
df_dm = pd.read_sql("SELECT * FROM dbo.clean_derived_metrics", cleaning_engine)
df_pl = pd.read_sql("SELECT * FROM dbo.clean_profit_loss",     cleaning_engine)

frames = {
    "clean_balance_sheet"   : df_bs,
    "clean_cash_flow"       : df_cf,
    "clean_derived_metrics" : df_dm,
    "clean_profit_loss"     : df_pl,
}

df_all = pd.concat(
    [df_bs, df_cf, df_dm, df_pl],
    ignore_index=True
)

df_all["Date"] = pd.to_datetime(df_all["Date"])
df_all["Date"] = df_all["Date"].dt.date

df_all.to_sql(
    name      = "clean_all_financials",
    con       = cleaning_engine,
    if_exists = "replace",   
    index     = False,
    schema    = "dbo",
)
print(f"Written to cleaning.dbo.clean_all_financials  ({len(df_all):,} rows)")

```
